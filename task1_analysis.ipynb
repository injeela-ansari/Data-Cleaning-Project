```
import os
from typing import List, Dict, Tuple, Optional

def generate_dummy_file(filename: str, numbers: List[int]) -> None:
    """
    Helper function to create a text file with random numbers for demonstration.

    Args:
        filename (str): The name of the file to create.
        numbers (List[int]): The list of numbers to write to the file.
    """
    try:
        with open(filename, 'w') as f:
            for num in numbers:
                f.write(f"{num}\n")
        print(f"[INFO] Created dummy file: {filename}")
    except IOError as e:
        print(f"[ERROR] Failed to create file: {e}")

def calculate_manual_stats(data: List[float]) -> Dict[str, float]:
    """
    Calculates mean, median, min, and max manually without using NumPy.

    Args:
        data (List[float]): A list of numerical values.

    Returns:
        Dict[str, float]: A dictionary containing the calculated statistics.
    """
    if not data:
        return {}

    # 1. Calculate Min and Max
    min_val = data[0]
    max_val = data[0]
    total_sum = 0

    for num in data:
        if num < min_val:
            min_val = num
        if num > max_val:
            max_val = num
        total_sum += num

    # 2. Calculate Mean
    count = len(data)
    mean_val = total_sum / count

    # 3. Calculate Median
    sorted_data = sorted(data)
    mid_index = count // 2

    if count % 2 == 0:
        # Even number of elements: average of the two middle numbers
        median_val = (sorted_data[mid_index - 1] + sorted_data[mid_index]) / 2
    else:
        # Odd number of elements: the middle number
        median_val = sorted_data[mid_index]

    return {
        "mean": mean_val,
        "median": median_val,
        "min": min_val,
        "max": max_val
    }

def get_top_scorers(student_marks: Dict[str, int], top_n: int = 3) -> List[Tuple[str, int]]:
    """
    Identifies the top N scorers from a dictionary of students.

    Args:
        student_marks (Dict[str, int]): Dictionary with student names as keys and marks as values.
        top_n (int): The number of top scorers to retrieve.

    Returns:
        List[Tuple[str, int]]: A list of tuples containing (student_name, mark) for the top scorers.
    """
    # Sort the dictionary items by value (marks) in descending order
    sorted_students = sorted(student_marks.items(), key=lambda item: item[1], reverse=True)
    return sorted_students[:top_n]

def sum_numbers_from_file(filename: str) -> Optional[float]:
    """
    Reads a text file containing numbers (one per line) and calculates their sum.

    Args:
        filename (str): Path to the text file.

    Returns:
        Optional[float]: The sum of numbers, or None if an error occurs.
    """
    total_sum = 0.0
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
            for line in lines:
                stripped_line = line.strip()
                if stripped_line: # Ensure line is not empty
                    total_sum += float(stripped_line)
        return total_sum
    except FileNotFoundError:
        print(f"[ERROR] File '{filename}' not found.")
        return None
    except ValueError:
        print(f"[ERROR] File '{filename}' contains non-numeric data.")
        return None

if __name__ == "__main__":
    print("--- Task 1: Python Data Handling Execution ---\n")

    # --- Part 1: Manual Statistics (List of 30+ numbers) ---
    # Generating a list of 35 numbers for demonstration
    dataset = [
        45, 67, 23, 89, 12, 56, 78, 90, 34, 21,
        65, 87, 43, 19, 76, 54, 32, 10, 98, 72,
        33, 55, 77, 99, 11, 22, 44, 66, 88, 100,
        15, 25, 35, 45, 55
    ]

    stats = calculate_manual_stats(dataset)
    print(f"Dataset Size: {len(dataset)}")
    print("Calculated Statistics (Manual):")
    for key, value in stats.items():
        print(f"  - {key.capitalize()}: {value:.2f}")

    print("-" * 30)

    # --- Part 2: Dictionary of Students & Top Scorers ---
    students = {
        "Aarav": 85, "Vihaan": 92, "Aditya": 78, "Sai": 88,
        "Reyansh": 95, "Arjun": 67, "Vivaan": 74, "Shaurya": 81,
        "Rohan": 90, "Ishaan": 89
    }

    top_3 = get_top_scorers(students, 3)
    print("\nTop 3 Scorers:")
    for rank, (name, score) in enumerate(top_3, 1):
        print(f"  {rank}. {name} - {score}")

    print("-" * 30)

    # --- Part 3: File Handling ---
    filename = "random_numbers.txt"

    # Generate the file first so the script is self-contained and runnable
    generate_dummy_file(filename, [10, 20, 30, 40, 50, 100, 200, 500])

    file_sum = sum_numbers_from_file(filename)
    if file_sum is not None:
        print(f"\nSum of numbers from '{filename}': {file_sum}")

    # Optional cleanup
    # os.remove(filename)
```
