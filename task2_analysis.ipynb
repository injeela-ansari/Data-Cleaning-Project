```python
import pandas as pd
import numpy as np
import os

# --- Step 1: Generate a Dummy Dataset (Minimum 500 rows) ---
# We will create a synthetic 'Sales' dataset for demonstration purposes.
def create_dummy_csv(filename='sales_data.csv', rows=600):
    np.random.seed(42)  # Ensure reproducibility

    categories = ['Electronics', 'Clothing', 'Home & Garden', 'Automotive', 'Sports']
    data = {
        'OrderID': range(1, rows + 1),
        'Category': np.random.choice(categories, rows),
        'Price': np.random.uniform(10, 1000, rows).round(2),
        'Quantity': np.random.randint(1, 20, rows),
        'CustomerRating': np.random.uniform(1, 5, rows).round(1)
    }

    df = pd.DataFrame(data)
    df.to_csv(filename, index=False)
    print(f"Successfully created '{filename}' with {rows} rows.")

# Create the file if it doesn't exist
if not os.path.exists('sales_data.csv'):
    create_dummy_csv()

# --- Step 2: Load the Dataset ---
df = pd.read_csv('sales_data.csv')

# --- Step 3: Display Basic Information ---
print("--- First 10 Rows ---")
print(df.head(10))

print("\n--- Last 10 Rows ---")
print(df.tail(10))

print(f"\nDataset Shape: {df.shape}")

# --- Step 4: Create 3 New Calculated Columns ---
# 1. TotalSales: Price multiplied by Quantity
df['TotalSales'] = df['Price'] * df['Quantity']

# 2. DiscountedPrice: Apply a flat 10% discount to the Price
df['DiscountedPrice'] = df['Price'] * 0.90

# 3. HighValueOrder: Boolean flag if TotalSales exceeds 500
df['HighValueOrder'] = df['TotalSales'] > 500

print("\n--- Added Calculated Columns (First 5 rows) ---")
print(df[['OrderID', 'TotalSales', 'DiscountedPrice', 'HighValueOrder']].head())

# --- Step 5: Filter Rows based on 2 Conditions ---
# Condition: Category is 'Electronics' AND CustomerRating is greater than 4.0
high_rated_electronics = df[
    (df['Category'] == 'Electronics') &
    (df['CustomerRating'] > 4.0)
]

print(f"\nfiltered Rows (Electronics & Rating > 4.0): {len(high_rated_electronics)}")

# --- Step 6: Group Data and Aggregate ---
# Group by 'Category' and calculate the mean of 'TotalSales' and the count of orders
category_stats = df.groupby('Category').agg({
    'TotalSales': 'mean',
    'OrderID': 'count'
}).rename(columns={'TotalSales': 'AverageSales', 'OrderID': 'OrderCount'})

print("\n--- Aggregated Statistics by Category ---")
print(category_stats)
```
